```python
import warnings
warnings.filterwarnings('ignore')

import pandas as pd
import numpy as np
```

```python
data = pd.read_csv('df_data.csv').dropna()

data
```

```python
global_mean = data['offset'].mean()
global_std = data['offset'].std()

threshold = 3 * global_std

print(threshold)
```

```python
def filter_outliers(df):
    station_mean = df['offset'].mean()
    return df[(df['offset'] >= station_mean - threshold) & (df['offset'] <= station_mean + threshold)]

filtered_data= data.groupby('station_id').apply(filter_outliers).reset_index(drop=True)
```

```python
filtered_data
```

```python
print(data['station_id'].nunique())
print(filtered_data['station_id'].nunique())
```

```python
stations_dict= (filtered_data['station_id'].value_counts()).to_dict()

stations_dict
```

```python
count_167 = sum(value == 167 for value in stations_dict.values())

print(f"The number of dictionary elements with value 167 is: {count_167}")
```
